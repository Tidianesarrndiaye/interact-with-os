## Core Methods in Subprocess

All methods—`.run()`, `.call()`, `.check_output()`, and `.check_call()`—are wrappers around the foundational `.Popen` class.

### 1. `.run()` (Recommended)

Runs a command, waits for it to complete, and returns a `CompletedProcess` instance containing execution details.

````python
result_run = subprocess.run(['echo', 'Hello, World!'], capture_output=True, text=True)
print(result_run.stdout.strip())
# Output: 'Hello, World!'
````

### 2. `.call()` (Legacy)

Runs a command, waits for completion, and returns the integer return code. `.run()` is preferred for modern code.

````python
return_code_call = subprocess.call(['echo', 'Hello from call!'])
# Output: 0 (indicates successful execution)
````

### 3. `.check_call()` and `.check_output()`

Ideal for situations like file I/O where operations might fail. Both raise a `CalledProcessError` exception if the command returns a non-zero exit code.

* **`check_call()`**: Returns just the status code.

````python
return_code_check_call = subprocess.check_call(['echo', 'Hello from check_call!'])
# Output: 0
````

* **`check_output()`**: Returns the command's standard output.

````python
output_check_output = subprocess.check_output(['echo', 'Hello from check_output!'], text=True)
print(output_check_output.strip())
# Output: 'Hello from check_output!'
````

### 4. `.Popen()` (Advanced)

The most fully featured way to call external commands. It spawns a new process, connects to input/output/error pipes, and obtains return codes. It is particularly powerful for **asynchronous behavior** and inter-process communication.

````python
process_popen = subprocess.Popen(['echo', 'Hello from popen!'], stdout=subprocess.PIPE, text=True)
output_popen, _ = process_popen.communicate()
print(output_popen.strip())
# Output: 'Hello from popen!'
````

---

## Pro Tip: Asynchronous Behavior with Popen

Use `Popen` when you need to start a long-running command in the background, continue with other tasks, and check on its status later using `.poll()`.

````python
import subprocess
import time

# Start a long-running process in the background (sleeps for 5 seconds)
process = subprocess.Popen(['sleep', '5'])
print("The process is running in the background...")

# Give it a couple of seconds to demonstrate asynchronous behavior
time.sleep(2)

# Check if the process has finished
if process.poll() is None:
    message = "The process is still running."
else:
    message = "The process has finished."

print(message)
# Output: 'The process is still running.' (since poll was checked after 2s of a 5s sleep)
````
